```python
from batch import *

# EV and Ru(bpy)3 stock solutions now made in pH 6 buffer
# mixing limitations in cuvette, colour gradient seen

# V(Ru(bpy)3, 1 mM, buffer) = 200 ul, V(EDTA, saturated, buffer) = 1 ml, made up to 3 ml pH 6 buffer
# V(EV, 1 mM, buffer) in ul used and start times for each run, start time adjusted slightly to begin at linear region
# blue 455 nm irradiation

x = 'E'
duration = 15*60
root = f"./results/260630_Rubpy_{x}V_EDTA_Buffered/Rubpy_{x}V_"
runs = [(300, "12-11-37-000"), # 12-10-37-000 recorded start time
        (600, "12-52-20-000"), # 12-51-20-000
        (900, "13-35-15-000"), # 13-35-15-000
        (1200, "14-10-27-000"), # 14-10-27-000
        (1500, "14-45-38-000"), # 14-45-38-000
        (1800, "15-19-52-000")] # 15-19-52-000

for (v, start) in runs:
    data = f'{root}{v}_EDTA_Buffered'
    concvstimeplot(data, v, start, x, duration=duration,
                   istart=100, r2threshold=0.99, interceptthreshold=1e-6)

for (v, start) in runs:
    data = f'{root}{v}_EDTA_Buffered'
    c0 = solprop(cstock=1e-3, valiquot=v*1e-6, vsol=3e-3) # initial concentration
    lncvstimeplot(data, v, start, x, duration=duration,
                  istart=100, r2threshold=0.99, interceptthreshold=np.log(c0))

for (v, start) in runs:
    data = f'{root}{v}_EDTA_Buffered'
    spectrumvstimeplot(data, v, start, x, duration=duration, vmin=0, vmax=1)

spectrumvsconcplot(root, runs, x, t=5*60)
```
